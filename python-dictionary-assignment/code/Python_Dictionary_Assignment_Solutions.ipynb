## Q1. What is a dictionary in Python?

A dictionary is a collection of **key-value pairs**.

- Keys are unique.
- Values can be of any data type.
- Dictionaries are written using `{}`.


````python
student = {
    "Name": "John",
    "Age": 21,
    "Department": "CSE"
}

print(student)
````

## Q4. What happens if you access a key that does not exist?

Using `student["Marks"]` raises a **KeyError**.

A safer alternative is `student.get("Marks)`, which returns `None` if the key does not exist.


````python
# This would raise KeyError:
# print(student["Marks"])

print(student.get("Marks"))
````

## Q5. Difference between `student["Age"]` and `student.get("Age")`

- `student["Age"]` returns the value if the key exists, otherwise raises `KeyError`.
- `student.get("Age")` returns the value if the key exists, otherwise returns `None`.
- `get()` can also accept a default value.


````python
print(student["Age"])
print(student.get("Age"))
print(student.get("Marks", 0))
````

## Q36. Count the frequency of each character in a string

````python
text = "banana"

frequency = {}

for char in text:
    frequency[char] = frequency.get(char, 0) + 1

print(frequency)
````

## Q37. Count the frequency of words in a sentence

````python
sentence = "I love Python and I love AI"

words = sentence.split()
frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

print(frequency)
````

## Q38. Reverse a dictionary

This approach assumes that the values are unique. If duplicate values exist, later keys will overwrite earlier keys.


````python
data = {"A": 1, "B": 2}

reversed_dict = {}

for key, value in data.items():
    reversed_dict[value] = key

print(reversed_dict)
````

## Q39. Find duplicate values in a dictionary

````python
data = {
    "A": 10,
    "B": 20,
    "C": 10,
    "D": 30,
    "E": 20
}

value_counts = {}

for value in data.values():
    value_counts[value] = value_counts.get(value, 0) + 1

duplicates = [value for value, count in value_counts.items() if count > 1]

print("Duplicate values:", duplicates)
````

## Q55. Create a Phonebook

The program supports:

1. Add a contact
2. Search for a contact
3. Update a contact
4. Delete a contact
5. Display all contacts
6. Exit


````python
phonebook = {}

while True:
    print("\n===== PHONEBOOK =====")
    print("1. Add Contact")
    print("2. Search Contact")
    print("3. Update Contact")
    print("4. Delete Contact")
    print("5. Display All Contacts")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        name = input("Enter name: ")
        phone = input("Enter phone number: ")

        phonebook[name] = phone
        print("Contact added successfully.")

    elif choice == "2":
        name = input("Enter name to search: ")

        if name in phonebook:
            print("Phone:", phonebook[name])
        else:
            print("Contact not found.")

    elif choice == "3":
        name = input("Enter name to update: ")

        if name in phonebook:
            phone = input("Enter new phone number: ")
            phonebook[name] = phone
            print("Contact updated successfully.")
        else:
            print("Contact not found.")

    elif choice == "4":
        name = input("Enter name to delete: ")

        if name in phonebook:
            del phonebook[name]
            print("Contact deleted successfully.")
        else:
            print("Contact not found.")

    elif choice == "5":
        if not phonebook:
            print("Phonebook is empty.")
        else:
            print("\nContacts:")
            for name, phone in phonebook.items():
                print(name, ":", phone)

    elif choice == "6":
        print("Exiting phonebook...")
        break

    else:
        print("Invalid choice. Please try again.")
````

# Quick Revision – Important Dictionary Methods

| Method | Purpose |
|---|---|
| `keys()` | Returns all keys |
| `values()` | Returns all values |
| `items()` | Returns key-value pairs |
| `get()` | Safely gets a value |
| `pop()` | Removes a specified key |
| `popitem()` | Removes the last inserted pair |
| `update()` | Adds/updates dictionary items |
| `setdefault()` | Gets a value or creates a default |
| `fromkeys()` | Creates a dictionary from keys |
| `clear()` | Removes all items |
| `copy()` | Creates a shallow copy |

### Three patterns to remember

**Frequency counting**
```python
frequency[key] = frequency.get(key, 0) + 1
```

**Find key with maximum value**
```python
key = max(dictionary, key=dictionary.get)
```

**Dictionary comprehension**
```python
result = {key: value for key, value in something}
```
