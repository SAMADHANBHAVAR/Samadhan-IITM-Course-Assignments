## Load the JSON file

### Important: Employee IDs from JSON

Because JSON object keys are strings, use:

```python
employees["1050"]
```

not:

```python
employees[1050]
```

## Q30. Generate a report showing Total Employees, Missing Email, Missing Phone, Average Salary, Highest Salary, and Lowest Salary.

````python
total_employees = len(employees)

missing_email_count = sum(
    1 for employee in employees.values()
    if employee.get("Email") is None
)

missing_phone_count = sum(
    1 for employee in employees.values()
    if employee.get("Phone") is None
)

average_salary = sum(
    employee["Salary"] for employee in employees.values()
) / total_employees

highest_employee = max(
    employees.values(),
    key=lambda employee: employee["Salary"]
)

lowest_employee = min(
    employees.values(),
    key=lambda employee: employee["Salary"]
)

print("========== EMPLOYEE REPORT ==========")
print("Total Employees       :", total_employees)
print("Employees Missing Email:", missing_email_count)
print("Employees Missing Phone:", missing_phone_count)
print("Average Salary        :", round(average_salary, 2))
print("Highest Salary        :", highest_employee["Salary"])
print("Lowest Salary         :", lowest_employee["Salary"])
````

## Q36. Find the department with the highest average salary.

````python
department_salaries = {}

for employee in employees.values():
    department = employee["Department"]

    if department not in department_salaries:
        department_salaries[department] = []

    department_salaries[department].append(employee["Salary"])

department_average_salary = {
    department: sum(salaries) / len(salaries)
    for department, salaries in department_salaries.items()
}

highest_avg_department = max(
    department_average_salary,
    key=department_average_salary.get
)

print("Department with highest average salary:", highest_avg_department)
print(
    "Average salary:",
    round(department_average_salary[highest_avg_department], 2)
)

print("\nAll department averages:")
for department, average in sorted(department_average_salary.items()):
    print(department, ":", round(average, 2))
````

## Q38. Find duplicate phone numbers, if any.

````python
phone_to_employees = {}

for employee in employees.values():
    phone = employee.get("Phone")

    if phone is not None:
        phone_to_employees.setdefault(phone, []).append(employee["Name"])

duplicates = {
    phone: names
    for phone, names in phone_to_employees.items()
    if len(names) > 1
}

if duplicates:
    print("Duplicate phone numbers:")
    for phone, names in duplicates.items():
        print(phone, "->", names)
else:
    print("No duplicate phone numbers found.")
````

# Quick Revision – Dictionary Concepts Used in Assignment 5

### Membership
```python
if key in dictionary:
    ...
```

### Safe lookup
```python
dictionary.get(key)
```

### Loop through key-value pairs
```python
for key, value in dictionary.items():
    ...
```

### Count values using a dictionary
```python
counts[key] = counts.get(key, 0) + 1
```

### Find maximum using a dictionary value
```python
max(dictionary, key=dictionary.get)
```

### Dictionary comprehension
```python
result = {
    key: value
    for key, value in dictionary.items()
    if condition
}
```
