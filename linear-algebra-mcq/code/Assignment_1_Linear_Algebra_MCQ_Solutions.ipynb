## Q1

**Question:**  A system is consistent if:

### ✅ Correct Answer
**B) rank(A) = rank([A|B])**

### Explanation
A system is consistent when the coefficient matrix and augmented matrix have the same rank.


## Q2

**Question:**  If rank(A) = rank([A|B]) = number of variables, the system has:

### ✅ Correct Answer
**C) Unique solution**

### Explanation
When the common rank equals the number of variables, there are no free variables, so the solution is unique.


## Q3

**Question:**  A homogeneous system always has:

### ✅ Correct Answer
**B) At least one solution**

### Explanation
A homogeneous system AX = 0 always has the trivial solution X = 0.


## Q4

**Question:**  If a homogeneous system has rank < number of variables, then:

### ✅ Correct Answer
**B) Infinite solutions**

### Explanation
A homogeneous system is always consistent. Rank less than the number of variables leaves free variables, giving infinitely many solutions.


## Q5

**Question:**  Which condition implies inconsistency?

### ✅ Correct Answer
**B) rank(A) ≠ rank([A|B])**

### Explanation
Different ranks for A and the augmented matrix indicate inconsistency.


## Q6

**Question:**  If det(A) ≠ 0, then:

### ✅ Correct Answer
**C) A is invertible**

### Explanation
A square matrix with non-zero determinant is invertible.


## Q7

**Question:**  Eigenvalues are roots of:

### ✅ Correct Answer
**B) det(A - λI) = 0**

### Explanation
Eigenvalues are roots of the characteristic equation det(A − λI) = 0.


## Q8

An eigenvalue λ with eigenvector X satisfies AX = λX, where X is non-zero.

# Quick Revision

## Consistency
For `AX = B`:

- **Consistent:** `rank(A) = rank([A|B])`
- **Inconsistent:** `rank(A) != rank([A|B])`

## Number of solutions
For a consistent system with `n` variables:

- `rank(A) = n` → **Unique solution**
- `rank(A) < n` → **Infinite solutions**

If the ranks differ → **No solution**.

## Homogeneous system
`AX = 0` always has the trivial solution `X = 0`.

- rank = number of variables → only trivial solution
- rank < number of variables → infinitely many solutions

## Determinant
- `det(A) != 0` → invertible
- `det(A) = 0` → singular, no inverse

## Eigenvalues
`det(A - λI) = 0`

and

`AX = λX`

## Trace
Trace = sum of diagonal elements = sum of eigenvalues (with multiplicity).

## Important exam point
For questions saying **rank = number of variables** or **rank < number of variables**, always check consistency first. Rank alone does not distinguish a unique solution from an inconsistent system.
