## Q21. Find the project with the highest funding.

````python
highest_project = None
highest_funding = 0

for university in university_system["universities"].values():
    for faculty in university["faculties"].values():
        for department in faculty["departments"].values():
            if "labs" in department:
                for lab in department["labs"].values():
                    for project_name, project in lab.get("research_projects", {}).items():
                        if project["funding"] > highest_funding:
                            highest_funding = project["funding"]
                            highest_project = project_name

print("Highest Funded Project:", highest_project)
print("Funding:", highest_funding)
````

## Q25. Count total number of researchers in GTU_India.

````python
gtu_india = university_system["universities"]["GTU_India"]

researchers = set()

for faculty in gtu_india["faculties"].values():
    for department in faculty["departments"].values():
        if "labs" in department:
            for lab in department["labs"].values():
                for project in lab.get("research_projects", {}).values():
                    researchers.update(project["researchers"])

print("Total Researchers:", len(researchers))
print("Researchers:", researchers)
````

## Q26. Find which lab has the maximum funding.

````python
lab_funding = {}

for university in university_system["universities"].values():
    for faculty in university["faculties"].values():
        for department in faculty["departments"].values():
            if "labs" in department:
                for lab_name, lab in department["labs"].items():
                    total = sum(
                        project["funding"]
                        for project in lab.get("research_projects", {}).values()
                    )
                    lab_funding[lab_name] = total

max_lab = max(lab_funding, key=lab_funding.get)

print("Lab with maximum funding:", max_lab)
print("Total funding:", lab_funding[max_lab])
````

## Q27. Print projects grouped by status (ongoing/completed).

````python
projects_by_status = {
    "ongoing": [],
    "completed": []
}

for university in university_system["universities"].values():
    for faculty in university["faculties"].values():
        for department in faculty["departments"].values():
            if "labs" in department:
                for lab in department["labs"].values():
                    for project_name, project in lab.get("research_projects", {}).items():
                        status = project["status"]

                        if status not in projects_by_status:
                            projects_by_status[status] = []

                        projects_by_status[status].append(project_name)

for status, projects in projects_by_status.items():
    print("\n" + status.upper())
    print("-" * len(status))

    for project in projects:
        print(project)
````

## Q30. Write a function that searches for a project name and prints its full details.

````python
def search_project(project_name):
    for university_name, university in university_system["universities"].items():
        for faculty_name, faculty in university["faculties"].items():
            for department_name, department in faculty["departments"].items():

                if "labs" in department:
                    for lab_name, lab in department["labs"].items():

                        projects = lab.get("research_projects", {})

                        if project_name in projects:
                            print("Project Found!")
                            print("--------------------")
                            print("Project:", project_name)
                            print("University:", university_name)
                            print("Faculty:", faculty_name)
                            print("Department:", department_name)
                            print("Lab:", lab_name)
                            print("Details:", projects[project_name])
                            return

    print("Project not found.")


# Test the function
search_project("DeepVision")
````

# Quick Revision – Nested Dictionary Navigation

### Direct access
```python
data["key1"]["key2"]["key3"]
```

### Loop through keys
```python
for key in data:
    print(key)
```

### Loop through values
```python
for value in data.values():
    print(value)
```

### Loop through key-value pairs
```python
for key, value in data.items():
    print(key, value)
```

### Safe access
```python
data.get("key")
```

### Find maximum
```python
max(data, key=data.get)
```

### Use a set for unique researchers
```python
researchers = set()
researchers.update(list_of_researchers)
```

### Important idea

For a deeply nested dictionary, first identify the path:

`university → faculty → department → lab → project → attribute`

Then either access the path directly or use nested loops when the level is dynamic.
